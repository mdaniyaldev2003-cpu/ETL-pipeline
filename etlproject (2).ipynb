```python
df = df.drop_duplicates()
```

```python
df.rename(columns={
    'Port Name': 'port_name',
    'State': 'state',
    'Port Code': 'port_code',
    'Border': 'border',
    'Date': 'date',
    'Measure': 'measure',
    'Value': 'value',
    'Latitude': 'latitude',
    'Longitude': 'longitude',
    'Point': 'point'
}, inplace=True)
```

```python
df['date'] = pd.to_datetime(df['date'], format='%b %Y', errors='coerce')
```

```python
df = df.dropna(subset=['date'])
```

```python
df['port_code'] = pd.to_numeric(df['port_code'], errors='coerce').fillna(0).astype(int)
df['value'] = pd.to_numeric(df['value'], errors='coerce').fillna(0).astype(int)
df['latitude'] = pd.to_numeric(df['latitude'], errors='coerce')
df['longitude'] = pd.to_numeric(df['longitude'], errors='coerce')
```

```python
df = df[['port_name', 'state', 'port_code', 'border', 'date', 'measure', 'value', 'latitude', 'longitude', 'point']]
```

```python
df.to_csv('border_cleaned.csv', index=False)
```
